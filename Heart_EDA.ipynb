# Outlier Detection

```python
numeric_cols = df.select_dtypes(include=np.number).columns

outlier_summary = {}

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    outliers = ((df[col] < lower) | (df[col] > upper)).sum()
    
    outlier_summary[col] = outliers

outlier_df = pd.DataFrame(
    outlier_summary.items(),
    columns=["Variable", "Outlier_Count"]
).sort_values(
    "Outlier_Count",
    ascending=False
)

outlier_df
```

#### "The dataset contains some statistical outliers. However, these observations were retained because extreme clinical values may represent genuine patient conditions and removing them could lead to loss of important information.

# Feature Engineering / Preprocessing

```python
df["age_group"] = pd.cut(
    df["age"],
    bins=[17, 30, 45, 60, 75, 100],
    labels=["18-30", "31-45", "46-60", "61-75", "76+"]
)
```

```python
df["age_group"].value_counts().sort_index()
```

# Age Group vs Heart Disease

```python
age_group_risk = (
    df.groupby("age_group", observed=False)["has_heart_disease"]
      .mean()
      .reset_index()
)

plt.figure(figsize=(8,5))

sns.barplot(
    data=age_group_risk,
    x="age_group",
    y="has_heart_disease"
)

plt.title("Heart Disease Rate by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Heart Disease Rate")
plt.show()
```

# Inference

Heart-disease prevalence generally becomes higher in older age groups, supporting the relationship between age and heart disease observed in the numerical analysis.
